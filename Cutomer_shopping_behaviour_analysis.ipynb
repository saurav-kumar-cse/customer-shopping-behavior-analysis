#### handle missing value with median over mean, because median is robust to outliers
#### handling missing value in Review Rating 37 rows with median can work technically, but this is not a good practice because rating can vary a lot depending on the category. review rating can be different for clothing category than footwear category.Now if we just take overalll median and fill it everywhere we basically forcing clothing itesm to get samee  review rating as footwearitems and that can introduce bias into our dataset. instead what will be best approach is to fill the missing review rating values using  median review rating within the each category.

```python
# Imputing missing values in Review Rating column with the median rating of the product category

df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
df
```

```python
#creaating a column age_group
df['age_group'] = pd.qcut(df['age'], q=4, labels=['Young Adult', 'Adult', 'Middle-Aged', 'Senior'])
```

```python
# create new column purchase_frequency_days

frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```
